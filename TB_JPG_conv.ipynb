```python
# abhishek code

import os
!pip install pydicom
import pydicom
import numpy as np
from PIL import Image


def convert_dicom_to_jpg(folder_a, folder_b):
    # Create folder B if it doesn't exist
    if not os.path.exists(folder_b):
        os.makedirs(folder_b)

    # Loop through all files in folder A
    for filename in os.listdir(folder_a):
        if filename.endswith(".dicom"):
            dicom_path = os.path.join(folder_a, filename)

            # Read the DICOM file
            dicom_data = pydicom.dcmread(dicom_path)

            # Get the pixel array from the DICOM file
            image_array = dicom_data.pixel_array

            # Convert the pixel array to an 8-bit image (if necessary)
            image_array = (np.maximum(image_array, 0) / image_array.max()) * 255.0
            image_array = np.uint8(image_array)

            # Convert to PIL Image
            image = Image.fromarray(image_array)

            # Save the image as JPG in folder B
            jpg_filename = os.path.splitext(filename)[0] + ".jpg"
            jpg_path = os.path.join(folder_b, jpg_filename)
            image.save(jpg_path)
            print(f"Converted {filename} to {jpg_filename}")

# Converting images from 'intbtr12' under 'NORMAL_DICOM' to jpg and storing it under 'JPG_NORMAL' folder.

folder_a = 'D:/intbtr257'
folder_b = 'D:/JPG_TB'

convert_dicom_to_jpg(folder_a, folder_b)

print('Conversion Completed.')
```
