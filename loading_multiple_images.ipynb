```python
import os
import cv2
import matplotlib.pyplot as plt

# Directory where images are stored
image_dir = 'Example'

# Function to load multiple images from directory
def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, filename))
        if img is not None:
            images.append(img)
    return images

# Load all images from directory
images = load_images_from_folder(image_dir)

# Display multiple images in a loop
for i, img in enumerate(images[:5]):  # Display the first 5 images
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB for correct display
    plt.figure()
    plt.imshow(img_rgb)
    plt.title(f"Image {i + 1}")
    plt.axis('off')
    plt.show()
```
